This line fills missing values (NaN) in the Review Rating column using the median review rating of each category. Example:- Electronics, clothing etc.

This code groups the dataframe by Category and selects the Review Rating column for each group.
Using transform(), it fills missing (NaN) review ratings with the median rating of that specific category while keeping the original dataframe structure intact.

```python
df['Review Rating'] =df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns
```

pd.qcut is used to divide the whole data set in quantiles and here we are dividing it in 4 parts with same number of observation.
pd.cut simply divides in 4 equal parts.

```python
#creating a new column age_group from age

labels = ['Young-Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[['age', 'age_group']].head(10)
```

```python
#create column purchase_frequency_days
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Bi-weekly': 14,
    'Monthly': 30,
    'Quaterly': 90,
    'Annually': 365,
    'Every 3 months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
# ================================
# Q1. Total Revenue by Gender
# ================================

query = """
SELECT gender,
       SUM(purchase_amount) AS revenue
FROM customers
GROUP BY gender
"""

result = pd.read_sql_query(query, conn)

result
```

```python
# ================================
# Q2. Customers Using Discount But Spending Above Average
# ================================

query = """
SELECT customer_id,
       purchase_amount
FROM customers
WHERE discount_applied = 'Yes'
AND purchase_amount >= (
    SELECT AVG(purchase_amount)
    FROM customers
)
"""

result = pd.read_sql_query(query, conn)

result
```

```python
# ================================
# Q3. Top 5 Products with Highest Average Review Rating
# ================================

query = """
SELECT item_purchased,
       ROUND(AVG(review_rating), 2) AS average_product_rating
FROM customers
GROUP BY item_purchased
ORDER BY AVG(review_rating) DESC
LIMIT 5
"""

result = pd.read_sql_query(query, conn)

result
```

```python
# ================================
# Q4. Average Purchase Amount:
# Standard vs Express Shipping
# ================================

query = """
SELECT shipping_type,
       ROUND(AVG(purchase_amount), 2) AS average_purchase
FROM customers
WHERE shipping_type IN ('Standard', 'Express')
GROUP BY shipping_type
"""

result = pd.read_sql_query(query, conn)

result
```

```python
# ================================
# Q5. Subscriber vs Non-Subscriber Spending
# ================================

query = """
SELECT subscription_status,
       COUNT(customer_id) AS total_customers,
       ROUND(AVG(purchase_amount), 2) AS avg_spend,
       ROUND(SUM(purchase_amount), 2) AS total_revenue
FROM customers
GROUP BY subscription_status
ORDER BY total_revenue DESC
"""

result = pd.read_sql_query(query, conn)

result
```

```python
# ================================
# Q6. Top 5 Products with Highest Discount Usage %
# ================================

query = """
SELECT item_purchased,
       ROUND(
           100.0 * SUM(
               CASE
                   WHEN discount_applied = 'Yes' THEN 1
                   ELSE 0
               END
           ) / COUNT(*),
       2) AS discount_rate
FROM customers
GROUP BY item_purchased
ORDER BY discount_rate DESC
LIMIT 5
"""

result = pd.read_sql_query(query, conn)

result
```

```python
# ================================
# Q7. Customer Segmentation
# ================================

query = """
WITH customer_type AS (
    SELECT customer_id,
           previous_purchases,
           CASE
               WHEN previous_purchases = 1 THEN 'New'
               WHEN previous_purchases BETWEEN 2 AND 10 THEN 'Returning'
               ELSE 'Loyal'
           END AS customer_segment
    FROM customers
)

SELECT customer_segment,
       COUNT(*) AS number_of_customers
FROM customer_type
GROUP BY customer_segment
"""

result = pd.read_sql_query(query, conn)

result
```

```python
# ================================
# Q8. Top 3 Products Within Each Category
# ================================

query = """
WITH item_counts AS (
    SELECT category,
           item_purchased,
           COUNT(customer_id) AS total_orders,
           ROW_NUMBER() OVER (
               PARTITION BY category
               ORDER BY COUNT(customer_id) DESC
           ) AS item_rank
    FROM customers
    GROUP BY category, item_purchased
)

SELECT item_rank,
       category,
       item_purchased,
       total_orders
FROM item_counts
WHERE item_rank <= 3
"""

result = pd.read_sql_query(query, conn)

result
```

```python
# ================================
# Q9. Repeat Buyers and Subscription Status
# ================================

query = """
SELECT subscription_status,
       COUNT(customer_id) AS repeat_buyers
FROM customers
WHERE previous_purchases > 5
GROUP BY subscription_status
"""

result = pd.read_sql_query(query, conn)

result
```

```python
# ================================
# Q9. Repeat Buyers and Subscription Status
# ================================

query = """
SELECT subscription_status,
       COUNT(customer_id) AS repeat_buyers
FROM customers
WHERE previous_purchases > 5
GROUP BY subscription_status
"""

result = pd.read_sql_query(query, conn)

result
```

```python
# ================================
# Q10. Revenue Contribution by Age Group
# ================================

query = """
SELECT age_group,
       SUM(purchase_amount) AS total_revenue
FROM customers
GROUP BY age_group
ORDER BY total_revenue DESC
"""

result = pd.read_sql_query(query, conn)

result
```
